TAP es la ventanilla para hacerle preguntas al catálogo y ADQL es el idioma de las preguntas. Crear `tap` todavía no pregunta nada.

## 3. Observaciones cerca de la estrella

Se busca por posición (círculo de 1 arcmin) y no por nombre, porque cada observador escribe el nombre distinto. Los nombres de columna (`s_ra`, `s_dec`) van sin `{}` y las variables de Python (`{ra}`, `{dec}`, `{radio}`) van con `{}`.

```python
radio = 1/60   # 1 arcmin, en grados

pregunta = f'''
SELECT dp_id, instrument_name, target_name, t_min, dataproduct_type, snr,
access_url
FROM ivoa.ObsCore
WHERE CONTAINS(POINT('ICRS', s_ra, s_dec),
               CIRCLE('ICRS', {ra}, {dec}, {radio})) = 1
'''

obs = tap.search(pregunta).to_table()
print(f'Productos encontrados: {len(obs)}')
```

## 4. Solo HARPS

Máscara de True/False: `&` es "y" y cada condición va entre paréntesis. Deben salir 296.

```python
HARPS = obs[(obs['instrument_name'] == 'HARPS') & (obs['dataproduct_type'] == 'spectrum')]
HARPS.sort('t_min')
print(f'Espectros de HARPS: {len(HARPS)}')
```

## 5. Función de descarga

- Si el archivo ya existe, no lo baja otra vez.
- Descarga de a 1 MB (`stream=True`) para no llenar la RAM.
- `raise_for_status()` frena si el servidor responde con error.
- Escribe en `.part` y renombra al final, así un archivo final siempre está completo.

```python
def descargar(url, destino):
    destino = Path(destino)
    if destino.exists():
        print(f'Ya existe, no lo bajo otra vez: {destino.name}')
        return destino
    destino.parent.mkdir(parents=True, exist_ok=True)
    temporal = destino.with_name(destino.name + '.part')

    with requests.get(url, stream=True, timeout=120) as r:
        r.raise_for_status()
        with open(temporal, 'wb') as f:
            for pedazo in r.iter_content(chunk_size=1024*1024):   # 1 MB
                f.write(pedazo)

    temporal.rename(destino)
    print(f'Descargado: {destino.name}')
    return destino
```

## 6. Archivos de un producto de HARPS

Cada producto tiene solo 2 archivos: el espectro 1D (`SCIENCE.SPECTRUM`) y un `.tar` (`ANCILLARY.HARPSTAR`) con los productos intermedios. El que sirve para LBL es el `.tar`.

```python
un_producto_harps = HARPS['dp_id'][0]
print(f'Producto: {un_producto_harps}')

archivos_harps = tap.search(f'''
SELECT archive_id, eso_category, original_filename, access_url, access_estsize
FROM phase3v2.product_files
WHERE product_id = '{un_producto_harps}'
''').to_table()

archivos_harps['eso_category', 'original_filename', 'access_estsize'].pprint_all()
```

## 7. Descargar y abrir un `.tar`

`tarfile.open` abre la caja sin sacar nada, `.getmembers()` lista lo que hay adentro (`.name`, `.size`) y `.extractfile()` lee un archivo sin guardarlo en disco.

Debe haber 11 archivos. El que usa LBL es `..._e2ds_A.fits` (4096 píxeles × 72 órdenes). El blaze **no** viene en el `.tar`.

```python
es_tar = [str(c) == 'ANCILLARY.HARPSTAR' for c in archivos_harps['eso_category']]
fila_tar = archivos_harps[es_tar][0]

ruta_tar = descargar(fila_tar['access_url'],
                     Path('datos_LHS_1140_TAR') / 'HARPS' / f"{fila_tar['archive_id']}.tar")

with tarfile.open(ruta_tar) as caja:
    for miembro in caja.getmembers():
        print(f'{miembro.size/1e6:8.2f} MB   {miembro.name}')
```

## 8. ¿Qué blaze usó este espectro?

El pipeline anota en el header del `e2ds` qué calibraciones usó. `extractfile(...).read()` da los bytes y `io.BytesIO` los envuelve para que `fits.open` los trate como un archivo.

```python
with tarfile.open(ruta_tar) as caja:
    m = [x for x in caja.getmembers() if x.name.endswith('e2ds_A.fits')][0]
    datos = caja.extractfile(m).read()

with fits.open(io.BytesIO(datos)) as hdul:
    hdul.info()
    h = hdul[0].header
    for clave in h:
        if any(s in clave.upper() for s in ['BLAZE', 'FLAT FILE', 'TH FILE']):
            print(clave, '=', h[clave])
```

## 9. Descargar todos los `.tar`

Ciclo sobre `HARPS`: para cada `dp_id` se consultan sus archivos, se deja la fila `ANCILLARY.HARPSTAR` y se baja el `.tar` (~7 MB cada uno, ~2 GB en total). En Colab todo se borra al cerrar la sesión, así que después se respalda en Drive (sección 12).

```python
for row in HARPS:
    # 1. Consultar los archivos de esta observación
    archivos_harps = tap.search(f'''
    SELECT archive_id, eso_category, access_url
    FROM phase3v2.product_files
    WHERE product_id = '{row['dp_id']}'
    ''').to_table()

    # 2. Filtrar el .tar
    es_tar = [str(c) == 'ANCILLARY.HARPSTAR' for c in archivos_harps['eso_category']]
    fila_tar = archivos_harps[es_tar][0]

    # 3. Descargar (tu función ya se encarga de ver si ya existe)
    descargar(fila_tar['access_url'],
              Path('datos_LHS_1140_TAR') / 'HARPS' / f"{fila_tar['archive_id']}.tar")
```

## 10. Sacar el `e2ds_A` de cada `.tar`

Se guarda en `science/LHS1140/`. Los `.tar` no traen el blaze, así que solo se guarda el espectro.

```python
# Carpeta de destino donde LBL buscará los datos listos
carpeta_destino = Path('science/LHS1140')
carpeta_destino.mkdir(parents=True, exist_ok=True)

# Ruta donde tienes tus .tar descargados
directorio_tar = Path('datos_LHS_1140_TAR') / 'HARPS'

# Recorres cada .tar
for ruta_tar in directorio_tar.glob('*.tar'):
    with tarfile.open(ruta_tar) as caja:

        # 1. Extraer el e2ds_A
        miembros_e2ds = [x for x in caja.getmembers() if x.name.endswith('e2ds_A.fits')]
        if not miembros_e2ds:
            continue
        miembro_e2ds = miembros_e2ds[0]
        datos_e2ds = caja.extractfile(miembro_e2ds).read()

        ruta_e2ds_salida = carpeta_destino / Path(miembro_e2ds.name).name
        with open(ruta_e2ds_salida, 'wb') as f:
            f.write(datos_e2ds)

    # 2. Extracción segura del blaze (solo si existe en la caja)
        miembros_blaze = [x for x in caja.getmembers() if x.name.endswith('_blaze_A.fits')]

        if miembros_blaze:
            miembro_blaze = miembros_blaze[0]
            datos_blaze = caja.extractfile(miembro_blaze).read()

            ruta_blaze_salida = carpeta_destino / Path(miembro_blaze.name).name
            with open(ruta_blaze_salida, 'wb') as f:
                f.write(datos_blaze)
            print(f"Guardados: {ruta_e2ds_salida.name} y {ruta_blaze_salida.name}")
        else:
            # Si es un paquete ADP o no trae blaze, guardamos solo el e2ds sin hacer drama
            print(f"Guardado solo e2ds (sin blaze interno): {ruta_e2ds_salida.name}")
```

## 14. Auditoría del blaze por época y descarte por S/N

Para cada espectro se calcula la correlación de Pearson entre el `e2ds` suavizado y el blaze, orden por orden. Si la correlación media baja, el blaze no le sirve a esa época. Luego se mira si eso depende de la S/N y se descartan los espectros con S/N(orden 47) < 5 (se mueven a `science/LHS1140_descartados/`, es reversible).

### Correlación por orden
Para cada espectro, `r` entre el `e2ds` suavizado y el blaze, orden por orden.

```python
carpeta_ciencia = Path('science/LHS1140')
calib = Path('calib')
ORDENES_DUDOSOS = [19, 43, 56, 64, 71]
ORDEN_MIN = 5          # LBL excluye los órdenes < 400 nm
VENTANA = 101          # píxeles de suavizado

def r_por_orden(e2ds, blaze):
    """Correlación de Pearson entre e2ds suavizado y blaze, orden por orden."""
    r = np.full(e2ds.shape[0], np.nan)
    for o in range(e2ds.shape[0]):
        y = np.nan_to_num(e2ds[o])
        if y.max() <= 0:
            continue
        ys = uniform_filter1d(y, VENTANA)
        r[o] = np.corrcoef(ys, blaze[o])[0, 1]
    return r

# --- 1. r por orden para cada espectro ---
filas = []
for ruta in sorted(carpeta_ciencia.glob('*_e2ds_A.fits')):
    with fits.open(ruta) as hdul:
        e2ds = np.array(hdul[0].data, dtype='float64')
        h = hdul[0].header
        nombre_blaze = h.get('ESO DRS BLAZE FILE')
        mjd = h.get('MJD-OBS')

    with fits.open(calib / nombre_blaze) as hb:
        blaze = np.array(hb[0].data, dtype='float64')

    r = r_por_orden(e2ds, blaze)
    fila = {'archivo': ruta.name, 'mjd': mjd}
    fila.update({f'o{o}': r[o] for o in range(len(r))})
    filas.append(fila)

df = pd.DataFrame(filas).sort_values('mjd').reset_index(drop=True)
cols_buenas = [f'o{o}' for o in range(ORDEN_MIN, 72)]
df['r_medio'] = df[cols_buenas].mean(axis=1)
df.to_csv('auditoria_blaze_por_epoca.csv', index=False)
print(f'{len(df)} espectros auditados')

# --- 2. r vs fecha ---
fig, ax = plt.subplots(1, 2, figsize=(13, 4))
ax[0].plot(df['mjd'], df['r_medio'], '.', ms=4)
ax[0].set_xlabel('MJD'); ax[0].set_ylabel(f'r medio (órdenes {ORDEN_MIN}-71)')
ax[0].set_title('¿Baja r con la fecha?')

for o in ORDENES_DUDOSOS:
    ax[1].plot(df['mjd'], df[f'o{o}'], '.', ms=3, label=f'orden {o}')
ax[1].set_xlabel('MJD'); ax[1].set_ylabel('r')
ax[1].set_title('Órdenes dudosos'); ax[1].legend(fontsize=8)
plt.tight_layout(); plt.savefig('auditoria_r_vs_fecha.png', dpi=130); plt.show()

# --- 3. Residuos e2ds/blaze en los órdenes dudosos (primer y último espectro) ---
fig, axs = plt.subplots(len(ORDENES_DUDOSOS), 2, figsize=(13, 2.4*len(ORDENES_DUDOSOS)),
                        sharex=True)
for j, idx in enumerate([0, len(df) - 1]):
    nombre = df.loc[idx, 'archivo']
    with fits.open(carpeta_ciencia / nombre) as hdul:
        e2ds = np.array(hdul[0].data, dtype='float64')
        nb = hdul[0].header.get('ESO DRS BLAZE FILE')
    with fits.open(calib / nb) as hb:
        blaze = np.array(hb[0].data, dtype='float64')

    for i, o in enumerate(ORDENES_DUDOSOS):
        res = np.nan_to_num(e2ds[o]) / blaze[o]
        res = res / np.nanmedian(res)                   # normalizado
        axs[i, j].plot(res, lw=0.4, color='0.6')
        axs[i, j].plot(uniform_filter1d(res, VENTANA), lw=1.2, color='C3')
        axs[i, j].set_ylabel(f'orden {o}')
    axs[0, j].set_title(nombre[:26], fontsize=9)
axs[-1, 0].set_xlabel('píxel'); axs[-1, 1].set_xlabel('píxel')
plt.tight_layout(); plt.savefig('auditoria_residuos.png', dpi=130); plt.show()
```

### ¿Depende de la S/N?
Se revisa si `r` baja cuando baja la S/N del orden 47.

```python
def leer_snr(ruta):
    with fits.open(ruta) as h:
        return h[0].header.get('ESO DRS SPE EXT SN47')

df['snr47'] = [leer_snr(carpeta_ciencia / a) for a in df['archivo']]

print(df.nsmallest(8, 'r_medio')[['archivo', 'mjd', 'r_medio', 'snr47']])

plt.figure(figsize=(6, 4))
plt.plot(df['snr47'], df['r_medio'], '.', ms=4)
plt.xlabel('S/N (orden 47)'); plt.ylabel('r medio')
plt.title('¿r baja cuando baja la S/N?')
plt.tight_layout(); plt.show()
```

### Descartar los espectros con S/N bajo
Se mueven a `science/LHS1140_descartados/` (reversible).

```python
UMBRAL_SNR = 5      # mismo umbral que SNR_THRESHOLD en el notebook de LBL

df['excluir'] = df['snr47'] < UMBRAL_SNR
print(f"Excluidos: {df['excluir'].sum()} de {len(df)}")
print(df[df['excluir']][['archivo', 'mjd', 'snr47', 'r_medio']])

# Mover los excluidos fuera de science/ (reversible)
descartados = Path('science/LHS1140_descartados')
descartados.mkdir(exist_ok=True)
for a in df.loc[df['excluir'], 'archivo']:
    (carpeta_ciencia / a).rename(descartados / a)

df.to_csv('auditoria_blaze_por_epoca.csv', index=False)
```

## 15. QC final

Revisión de lo que va a leer LBL: objeto, número de coeficientes de wave, versión del DRS, tipo de observación y S/N. Todo debe ser homogéneo en los 293 espectros.

```python
# =====================================================================
# QC de lo bajado, contra lo que LBL realmente lee
#
# Verificado contra el código fuente de LBL (njcuk9999/lbl, lbl/instruments/
# harps.py) y contra un e2ds_A real de LHS1140 descargado del archivo:
#   - longitud de onda: LBL la construye desde el propio header del e2ds
#     (288 claves únicas "ESO DRS CAL TH COEFF LL0..LL287"; el coeficiente
#     de grado j de la orden o es LL{4*o + j}; verificado: reconstruye
#     377.95-691.28 nm, monótona por orden). NO usa archivos wave_A:
#     el keyword "ESO DRS CAL TH FILE" solo se registra en la tabla RDB,
#     LBL nunca abre ese archivo.
#   - blaze: LBL lee "ESO DRS BLAZE FILE" del header del e2ds y busca un
#     archivo con ESE nombre exacto dentro del directorio calib/. Lo lee
#     con fits.getdata() sin extensión (encuentra datos aunque estén en
#     un HDU extendido, como en HARPS_BLAZE_A_72.fits).
#   - SNR: LBL usa "ESO DRS SPE EXT SN47" (orden 47) y por defecto descarta
#     SNR < 10; en el notebook de LBL el umbral se bajó a 5.
#
#   CLASE DE INSTRUMENTO: usar "HARPS" (la clase base), NO "HARPS_ESO".
#   HARPS_ESO es para los productos reprocesados del SDP (s1d/s2d ADP.*)
#   y activa BLAZE_CORRECTED=True (ignoraría los blazes de calib/) y lee
#   keywords "ESO QC *" que NO existen en estos e2ds_A del DRS.
#
# Antes de correr esto en Colab: monta Drive y apunta RAIZ a tu carpeta.
# =====================================================================

# ---------------------------------------------------------------------
# Celda 1: tabla de QC de los 296 e2ds
# ---------------------------------------------------------------------

RAIZ = Path('.')   # ojo: si trabajas directo desde Drive,
                   # usa Path('/content/drive/MyDrive/LHS1140_HARPS')

filas = []
for ruta in sorted((RAIZ / 'science' / 'LHS1140').glob('*_e2ds_A.fits')):
    with fits.open(ruta) as hdul:
        h = hdul[0].header
        # los coeficientes de la solución de longitud de onda
        coefs = [k for k in h if k.startswith('ESO DRS CAL TH COEFF LL')]
        filas.append({
            'e2ds': ruta.name,
            'objeto': h.get('ESO OBS TARG NAME'),        # LBL: KW_OBJNAME
            'dpr_type': h.get('ESO DPR TYPE'),           # LBL: KW_DPRTYPE
            'drs_version': h.get('ESO DRS VERSION'),
            'snr47': h.get('ESO DRS SPE EXT SN47'),      # LBL: umbral 10
            'n_coef_wave': len(coefs),                   # debe ser 288
            'blaze': h.get('ESO DRS BLAZE FILE'),        # LBL: lo busca en calib/
            'bjd': h.get('ESO DRS BJD'),                 # LBL: KW_MID_EXP_TIME
            'berv': h.get('ESO DRS BERV'),
            'exptime': h.get('ESO DET WIN1 DIT1'),
        })

qc = pd.DataFrame(filas)
qc.to_csv(RAIZ / 'qc_LHS1140.csv', index=False)

print('Objetos (debe ser solo LHS1140):')
print(qc['objeto'].value_counts().to_string())
print()
print('Coeficientes de wave por archivo (debe ser solo 288):')
print(qc['n_coef_wave'].value_counts().to_string())
print()
print('Versiones DRS:')
print(qc['drs_version'].value_counts(dropna=False).to_string())
print()
print('DPR TYPE:')
print(qc['dpr_type'].value_counts().to_string())
print()
print(f"SNR47: min={qc['snr47'].min():.1f}  mediana={qc['snr47'].median():.1f}  "
      f"max={qc['snr47'].max():.1f}")
print(f"Espectros con SNR47 < 10 (umbral de LBL): {(qc['snr47'] < 10).sum()}")
print(f"Espectros sin BJD o sin BERV: "
      f"{qc['bjd'].isna().sum() + qc['berv'].isna().sum()}")

# ---------------------------------------------------------------------
# Celda 2: cada e2ds debe encontrar SU blaze en calib/, con forma correcta
# (lectura idéntica a LBL: fits.getdata(ruta) sin extensión, que devuelve
#  el primer HDU con datos -- funciona aunque el blaze tenga los datos
#  en un HDU extendido "BLAZE" como HARPS_BLAZE_A_72.fits)
# ---------------------------------------------------------------------

calib = RAIZ / 'calib'
problemas = []

for blaze in qc['blaze'].unique():
    ruta = calib / blaze
    if not ruta.exists():
        problemas.append((blaze, 'NO EXISTE'))
        continue
    try:
        datos = fits.getdata(ruta)   # llamada exacta de lbl/core/io.load_fits
    except Exception as e:
        problemas.append((blaze, f'no se pudo leer: {e}'))
        continue
    if datos is None:
        problemas.append((blaze, 'fits.getdata no encontró datos'))
    elif datos.shape != (72, 4096):
        problemas.append((blaze, f'forma {datos.shape}, esperada (72, 4096)'))
    elif not np.isfinite(datos).all():
        n_nan = int((~np.isfinite(datos)).sum())
        problemas.append((blaze, f'{n_nan} píxeles no finitos (NaN/inf)'))

if problemas:
    print('PROBLEMAS:')
    for nombre, detalle in problemas:
        print(f'  {nombre}: {detalle}')
else:
    print(f"OK: los {qc['blaze'].nunique()} blazes existen, fits.getdata() los "
          f"lee con forma (72, 4096) y sin NaN/inf.")

# ---------------------------------------------------------------------
# Celda 3: resumen final (todo lo que LBL necesita antes de correr)
# ---------------------------------------------------------------------
chequeos = {
    'e2ds en science/LHS1140/': len(qc) == 296,
    'todos los archivos son LHS1140': (qc['objeto'] == 'LHS1140').all(),
    '288 coef. de wave en todos': (qc['n_coef_wave'] == 288).all(),
    'BJD presente en todos': qc['bjd'].notna().all(),
    'BERV presente en todos': qc['berv'].notna().all(),
    'blazes completos en calib/': len(problemas) == 0,
    'SNR47 >= 10 en todos': (qc['snr47'] >= 10).all(),
}
for nombre, ok in chequeos.items():
    print(f"{'OK ' if ok else '¡OJO!'}  {nombre}")
```
